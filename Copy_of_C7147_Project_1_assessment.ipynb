## 🧬 Load reference and reads data into variables
Below, we provide some helper code to load the reference from `reference.txt` and the reads from `reads.csv`.

❗ **Do not change the function’s behavior** (you may read it, of course!).

The code creates two variables from the loaded files:
- **reference**: a string containing the 200-nt reference sequence
- **reads**: a list of dictionaries containing the 6,000 reads

````python
##### Constants

REF_PATH = "reference.txt"
READS_PATH = "reads.csv"

##### Functions

def load_reference(path: str) -> str:
    """Load reference nucleotide sequence from file at path to reference string."""
    with open(path) as f:
        lines = [ln.strip() for ln in f if ln.strip()]
    if lines[0].startswith('>'):
        lines = lines[1:]
    ref = "".join(lines).upper()
    if len(ref) != 200:
        raise ValueError(f"Reference must be 200 nt, got {len(ref)}")
    return ref


def load_reads(path: str) -> list:
    """Load sample NGS reads from file at path to reads list."""
    with open(path, newline="") as f:
        reader = csv.DictReader(f)
        reads = list(reader)
    return reads

##### Main code

# Load reference and reads
reference = load_reference(REF_PATH)
reads = load_reads(READS_PATH)

# Report data structures
print("Reference length:", len(reference))
print("Total no. of reads:", len(reads))
print("Keys in each read dictionary:", reads[0].keys())
print("Samples loaded:", sorted({read["sample"] for read in reads}))

# Print sample sequences
print("\nSample read data:")
for row in reads[:5]:
    print(row["read_id"], row["sample"], row['group'], row["sequence"])
print("...")
for row in reads[-5:]:
    print(row["read_id"], row["sample"], row['group'], row["sequence"])
````

## Step 1 — Validate each read and record any errors
## Function: `validate_read(sequence, reference)`

### 👍 You can assume:
- `reference` is a non-empty string containing only uppercase A/C/G/T.
- `sequence` is a string, but nothing about its length or contents.

### 🎯 Aims:
- Check that an individual read is the same length as the reference and contains only valid characters (A/C/G/T)
- "Repair" any reads that contain valid lower-case characters -> upper-case
- Return a dictionary with the following keys:

    - `sequence`: the upper-case version of the sequence input, including when invalid.
    - `valid`: `True` only when the length matches the reference and every character is A/C/G/T.
    - `errors`: a list of strings. Include `"wrong_length"` first if lengths differ, then `"invalid_characters"` if any character is outside A/C/G/T. Include both when both apply. A valid read has `[]`.

### 🔎 Expected outputs

Examples for reference `"ATGC"`:
| Input (read) | Input type | Output |
|---|---|:---|
| `"ATGC"` | Correct | `{"sequence": "ATGC", "valid": True, "errors": []}` |
| `"atgc"` | Lowercase | `{"sequence": "ATGC", "valid": True, "errors": []}` |
| `""` | Empty read | `{"sequence": "", "valid": False, "errors": ["wrong_length"]}` |
| `"ATNC"` | Invalid character | `{"sequence": "ATNC", "valid": False, "errors": ["invalid_characters"]}` |
| `"ATN"` | Multiple errors | `{"sequence": "ATN", "valid": False, "errors": ["wrong_length", "invalid_characters"]}` |


Invalid reads should not crash your program.

````python

def validate_read(sequence, reference):
    """Checks reads for valid characters, and checks length is same as the reference. Also repairs lowercase reads."""

    errors = [] ## list where errors will be added

    seq_upper = sequence.upper() ## fixes lowercase sequences

    if len(sequence) != len(reference): ## checks if length is equal to reference if not error gets added to list
      errors.append("wrong_length")

    for base in seq_upper: ## checks if characters match between the bases and seq_upper
        if base not in "ATGC":
            errors.append("invalid_characters")
            break ## once error is found it stops the loop

    validation = { ## returns keys for the validation dictionary
        "sequence": seq_upper,
        "valid": len(errors) == 0,
        "errors": errors
    }
    return validation


    # Write your code here.
````

## Step 2 — Find every mismatch in a valid read
## Function: `find_mutations(sequence, reference)`

### 👍 You can assume:
- Inputs `sequence` and `reference` are valid (i.e. both are strings of the same length and contain only upper-case bases, i.e. A/C/G/T)

### 🎯 Aims:
- Return a list containing one dictionary per mismatch, in ascending position order.
- Each dictionary has exactly these keys:
    - `position`: position of the mutation (integer), numbered from 1.
    - `reference_base`: one-character string from reference at that position.
    - `read_base`: one-character string from the read at that position.

- Return `[]` if the read matches the reference.

### 🔎 Expected outputs

Example: `find_mutations("GTGA", "ATGC")` returns:
```python
[{"position": 1, "reference_base": "A", "read_base": "G"},
 {"position": 4, "reference_base": "C", "read_base": "A"}]
```


### 💻 Enter your code here

````python
def find_mutations(sequence, reference):
    """This function finds any mismatching bases between sequence and reference DNA, records their position and returns the data in a list."""
    mutations_list = [] ## creates empty list for any base mismatches to go into for a sequence
    for bases in range(len(reference)): ## creates index numbers for each character in the reference sequence
      position = bases + 1 ## the first characters index is 0 but we want to label it as position 1, this adds one to the original index number to shift the labels

      if sequence[bases] != reference[bases]: ## if sequence character doesn't match with reference character

        mutations = { ## the position, character reference, and sequence reference are put in in these dictionary keys
            "position": position,
            "reference_base": reference[bases],
            "read_base": sequence[bases]
        }
        mutations_list.append(mutations) ## collects all the found mismatches for a sequence

    return mutations_list ## gives back mismatches that were found

    # Write your code here.
````

## Step 3 — Analyse all reads
## Function: `analyse_reads(reads, reference)`

### 👍 You can assume:
- `reads` is a list of dictionaries containing keys `read_id`, `sample`, `group` and `sequence`, each with string values.
- `reference` is a non-empty string containing only A/C/G/T.

### 🎯 Aims:
- Use `validate_read` and `find_mutations` to return a **new list of dictionaries containing information about all reads** (even invalid ones, one dictionary per read), preserving the input read order.
- Each read dictionary should contain the following keys (values below are examples):
```python
{
    "read_id": "read_0001",
    "sample": "sample1",
    "group": "treatment",
    "sequence": "GTGA",             # normalised upper-case sequence
    "status": "valid",              # "valid" or "invalid"
    "errors": [],                   # validation error codes
    "mutations": [
        {"position": 1, "reference_base": "A", "read_base": "G"},
        {"position": 4, "reference_base": "C", "read_base": "A"}
    ]
}
```
- Do not change the original read dictionaries.

### 🔎 Expected outputs
- Valid reads have `status="valid"` and `mutations` listing all mutations in the dictionary format shown above.
- Invalid reads have `status="invalid"`, the relevant error codes, and `mutations=[]`.
- Empty input returns `[]`.

### 💻 Enter your code here

````python
def analyse_reads(reads, reference):
    """Returns all information about reads by using previous two functions and uses keys from the reads dictionary."""
    read_list = [] ## creates list so each read gets its own dict and not overwritten each time a read is looped through

    for read in reads:
      validation = validate_read(read["sequence"], reference) ## assigns a variable to the results of the validation function for a sequence

      if validation["valid"]:
         mismatches = find_mutations(validation["sequence"], reference) ## assigns a variable for the mutations function
      else:
        mismatches = [] ## if no mismatches were in that read, mismatches is kept empty

      result ={
          "read_id": read["read_id"],
          "sample": read["sample"],
          "group": read["group"],
          "sequence": validation["sequence"],
          "status": "valid" if validation["valid"] else "invalid", ## if statement which decides validity based on validation variable
          "errors": validation["errors"], ## again takes errors from the validation function
          "mutations": mismatches,
      }
      read_list.append(result) ## adds every read_id dictionary to the list
    return read_list
````

### 💻 Enter your code here

````python
def summarise_samples(analysed_reads):
  """this function takes analysed_reads data and sorts the reads into sample groups, counting their mismatches etc"""
  sample_summary = {}

  for read in analysed_reads:
    sample_number = read['sample'] ### assigns every sample to this variable

    if sample_number not in sample_summary: ### starts a fresh dictionary each time the for loop above comes across a new sample
      sample_summary[sample_number] = { ### all values start at 0, adds up everything for that sample, e.g. sample 1, sample 2
        "group": read['group'],
        "valid_reads": 0,
        "invalid_reads": 0,
        "mismatch_reads": 0,
        "mismatch_percentage": 0,
        "mismatch_events": 0,
        "substitution_counts": {
            "A>G":0, "A>C":0, "A>T":0,
            "C>T":0, "C>G":0, "C>A":0,
            "G>T":0, "G>A":0, "G>C":0,
            "T>G":0, "T>A":0, "T>C":0,
        },
        "position_counts": {},
        "top_positions": [],
       }
    summary = sample_summary[sample_number] ## makes it easier to type out to add values to the keys in the dict

    if read['status'] == 'valid': ### adds how many reads are valid
        summary['valid_reads'] += 1

        if len(read['mutations']) >= 1: ### adds to read mismatch count
            summary['mismatch_reads'] += 1
            summary['mismatch_events']+= len(read['mutations']) ### reads how many mutations occurred in the read and adds tha to event count

            for substitution in read['mutations']: ### sets up so that keys from mutations are stored in substitution
              position = substitution['position'] ### mutation position assigneed to variable position
              reference_base = substitution['reference_base']
              read_base = substitution['read_base']

              if position not in summary['position_counts']: ### if the mutation position is new its added to the count for that sample, if the counts already there it goes straight to adding to the count
                summary['position_counts'][position] = 0
              summary['position_counts'][position] += 1

              sub_mismatches = (f"{reference_base}>{read_base}") ### sorts the different mismatch substitutions into the substitution counts dict in sample summary
              if sub_mismatches in summary['substitution_counts']:
                summary['substitution_counts'][sub_mismatches] += 1

    else:
        summary['invalid_reads'] += 1

    ## after these loops with the final calculated positions the percentages for all 6 individual samples can be calculated and sorted into top 3
    for sample_number, summary in sample_summary.items():
      if summary['valid_reads'] >= 1:
        summary['mismatch_percentage'] = (summary['mismatch_reads'] / summary['valid_reads']) * 100 ### calculates mismatch percentage
      else:
        summary['mismatch_percentage'] = None ### if reads aren't valid percentage is returned as none
        print("No valid reads, percentage cannot be calculated")

      largest_positions = sorted(summary['position_counts'], key=summary['position_counts'].get, reverse = True)
      summary['top_positions'] = largest_positions[0:3]


  return sample_summary

def summary_tables (summarised):
  """this function produces two summary tables using data from the summarise function"""
  print("Summary table")
  print("-"*50)
  print(f"{'Sample':<10} | {'Group':<10} | {'Valid reads':<10} |{'Invalid reads':<10} | {'Mismatch percentage (%)':<12} | {'Mismatch events':<15} | {'Top positions'}")
  print (f"----------------------------------------------------------------------------------------------------------------") #### producing the headers of the table
  for sample_id, summary in summarised.items():
      if summary['mismatch_percentage'] is not None:
        percent = f"{summary['mismatch_percentage']:.2f}%"
      else:
        percent = "No mismatch percentage"
### below the row values are produced for each sample
      print(f"{sample_id:<10} | "
            f"{summary['group']:<10} | "
            f"{summary['valid_reads']:<11} | "
            f"{summary['invalid_reads']:<12} | "
            f"{percent:<23} |"
            f"{summary['mismatch_events']:<16} | "
            f"{summary['top_positions']}")
  print("-" * 50)
  print("Substitution summary table")
  print("-"*50)
### Below is substitution table
  column_names = f"{'Substitution':14} |"
  for sample_id in summarised:
    column_names += f'{sample_id:<10} count and (%)|' ## prints out sample names as the column headers
  print(column_names)
  print('-' * len(column_names))

  subs = ["A>G", "A>C", "A>T", ## replicates list of substitutions
            "C>T", "C>G", "C>A",
            "G>T", "G>A", "G>C",
            "T>G", "T>A", "T>C"]

  for sub in subs: ## creates row for a substitution
    row = f'{sub:<14} |'
    for sample_id in summarised:
      sub_counts = summarised[sample_id]['substitution_counts'][sub] ##selects the count of the mismatch for each sample
      valid_reads = summarised[sample_id]['valid_reads']

      if valid_reads >=1: ## checks for validity
        sub_percentage = (sub_counts/valid_reads)*100 ## percentage of that specific mismatch combo
        count_percent = f'{sub_counts} ({sub_percentage:.2f}%)'  ## puts the mismatch count together with its percentage so that it can be put in the table

      else:
        print("No percentage")

      row += f'{count_percent:<24}|' ## adds the count and percent of each sample for that specific substitution in the row
    print (row)


  return


````

## Step 5 — Exploratory comparison of the top mismatch positions

## Function:
`compare_positions(sample_summary, control_group="control", treated_group="treatment")`

### 👍 You can assume

- `sample_summary` is a dictionary in the format returned by the `summarise_samples` function
- `control_group` and `treated_group` are strings
- All samples have been analysed against the same reference sequence.

### 🎯 Aims

- Using the `position_counts` returned by `summarise_samples`, sum the mismatch counts at every position, **combining samples from `control_group` and `treatment_group`** (you should end up with a 1D array/list with the same length as a read).
- Select the **top 3 positions with the highest combined counts** (or fewer if there are less than three mismatches), choosing the earlier reference position in the case of a tie.
- For each position selected, calculate **mismatch percentages** separately for each biological sample.
- Compare treatment groups using the supplied statistical function `calculate_statistics` (see below).
- **Print out** the results of the statistical comparison clearly in standard scientific style, including appropriate statistical details based on the test used (a two-sided Welch's t-test).

🛠 **Parameters:**

- `sample_summary`: output from `summarise_samples` function.
- `control_group` and `treated_group`: labels identifying the groups to compare.

### 🧮 Sample-level calculations

For each selected position and sample, calculate:

**Mismatch percentage = valid reads with a mismatch at that position ÷ total valid reads × 100**

Count any substitution at the selected position. Include samples with zero mismatches, but exclude samples with no valid reads and report their exclusion.

Reuse helper functions from Step 4 where appropriate to reduce code complexity and repetition.

### 📊 Statistical comparison

For each position, pass the sample percentages to the supplied function as a list (one per group) and store the result, e.g.:

```python
statistical_result = calculate_statistics(control_percentages, treated_percentages)
```

Each list must contain **one percentage per biological sample**.

The supplied function returns:

```python
{"statistic": float, "p_value": float}
```

It returns `None` values when the comparison cannot be calculated.

````python
#@title 📊 calculate_statistics function

from scipy import stats
import math
import warnings


def calculate_statistics(control_percentages, treated_percentages):
    """Compare independent biological samples using a two-sided Welch's t-test."""
    unavailable = {"statistic": None, "p_value": None}

    # Require at least two samples per group and finite numerical values.
    if len(control_percentages) < 2 or len(treated_percentages) < 2:
        return unavailable

    values = list(control_percentages) + list(treated_percentages)
    if any(value is None or not math.isfinite(value) for value in values):
        return unavailable

    # No within-group variation: the t-test cannot estimate uncertainty.
    if (
        len(set(control_percentages)) == 1
        and len(set(treated_percentages)) == 1
    ):
        return unavailable

    with warnings.catch_warnings():
        warnings.simplefilter("ignore", RuntimeWarning)
        result = stats.ttest_ind(
            treated_percentages,
            control_percentages,
            equal_var=False,
            alternative="two-sided"
        )

    if not (
        math.isfinite(result.statistic)
        and math.isfinite(result.pvalue)
    ):
        return unavailable

    return {
        "statistic": float(result.statistic),
        "p_value": float(result.pvalue)
    }
````

### 📋 Return format

Return a list with one dictionary per selected position:

```python
[
    {
        "position": 42,

        "treated_mean": 2.0,
        "difference": 1.5,

    }
]
```

Means are sample mismatch percentages; `difference` is **treated mean − control mean**, in percentage points. Use `None` for results that cannot be calculated.

Print **one brief line per position**, reporting group means, their difference, sample numbers and p-value.

### 💻 Enter your code here

````python
# need to look at top positions across all 6 samples, then for each sample do the percentage of top positions
def compare_positions(sample_summary, control_group = "control", treated_group = "treatment"):
  """ This function sums up the counts of every mismatch position, narrows it down to 3 positions and then calculates mismatch percentages for each position across the 6 samples. It then preforms a two-sided welchs test on all 3 positions"""
  top_3_positions_overall = {} ## creates a dictionary to store the sum of every position of every sample

  for sample_id, summary in sample_summary.items(): ## this for loop adds the position counts for all six samples and puts them into their according group
    if summary["group"] in [control_group, treated_group]:
      for position, value in summary['position_counts'].items(): ## assigns the key as position and the counts in the value as 'value'
        if position not in top_3_positions_overall:
          top_3_positions_overall[position] = 0 ## the position is added as a key
          top_3_positions_overall[position] += value ## the value for that position is added to this dict
  sorted_top_3 = sorted(top_3_positions_overall, key = top_3_positions_overall.get, reverse = True ) ## sorts total counts of each position in this dictionary from high to low
  top_3 = sorted_top_3[0:3] ## stores a slice of the sorted variable into a new variable

  position_dict_list = [] ### sets up a list so there should be a entry for the 3 positions after each loop

  for position in top_3: ### initiates loop that cycles through the 3 positions and calculates percentages for each sample for them
    treatment_percentage = [] ## empty lists for the welch_test calculations
    control_percentage = []

    for sample_id, summary in sample_summary.items(): ##if the read is valid the position and its sum count are retrieved and calculated with a samples reads
      if summary["valid_reads"] >= 1:
        total_position_count = summary['position_counts'].get(position, 0)
        top_percentage = (total_position_count / summary["valid_reads"]) * 100

        if summary['group'] == treated_group: ## this if statement adds the result to the percentage list for the designated group
          treatment_percentage.append(top_percentage)
        elif summary['group'] == control_group:
          control_percentage.append(top_percentage)
      else:
        print(f"Can't be calculated, read is invalid")

    welch_test = calculate_statistics(control_percentage, treatment_percentage) ## storing function in the welch_test variable

    control_n = len(control_percentage) ## number of samples used in the sample size
    treated_n = len(treatment_percentage)
    control_mean = sum(control_percentage) / control_n
    treated_mean = sum(treatment_percentage) / treated_n
    difference = treated_mean - control_mean

    if welch_test is not None:
       t_stat = welch_test["statistic"] ## retrieves the results from the function to get the t statistic
       p_value = welch_test["p_value"]
    else:
        t_stat = None ## otherwise can't be calculated none
        p_value = None

    positions = { ## dictionary to store data for each of the 3 positions
         "position": position,
         "control_n": control_n,
         "treated_n": treated_n,
         "control_mean": control_mean,
         "treated_mean": treated_mean,
         "difference": difference,
         "statistic": t_stat,
         "p_value": p_value
     }
    position_dict_list.append(positions)

  for results in position_dict_list: ### this for loop just prints out the welchs test results for all 3 positions
    p_v = results['p_value']
    significance = "Significant" if (p_v is not None and p_v <= 0.05) else "Not significant"

    if results['statistic'] is not None: ## prevents program from just crashing, added after testing with a dummy data set
      print(f"Position: {results['position']}, t-statistic = {results['statistic']:.3f}, p_value = {p_v}, {significance}")
    else:
        print(f"Position: {results['position']}, t-statistic = None, {significance}")
  return position_dict_list

````
